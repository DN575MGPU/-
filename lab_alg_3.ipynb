Задача 18. Сортировка Шелла
Реализуйте функцию shell_sort(arr). Начальный шаг (gap) должен быть
равен len(arr) // 2 и уменьшаться вдвое на каждой итерации (целочисленное
деление).

```python
'''Тестовые данные:
a = [12, 34, 54, 2, 3]; shell_sort(a); print(a) # Ожидается: [2, 3, 12, 34, 54]
a = [5, 4, 3, 2, 1]; shell_sort(a); print(a) # Ожидается: [1, 2, 3, 4, 5]'''

def shell_sort(arr):
    # Сортировка Шелла с шагом n // 2
    n = len(arr)
    gap = n // 2

    while gap > 0:
        # Сортировка вставками с текущим шагом gap
        for i in range(gap, n):
            temp = arr[i]
            j = i
            # Сдвигаем элементы, находящиеся на расстоянии gap
            while j >= gap and arr[j - gap] > temp:
                arr[j] = arr[j - gap]
                j -= gap
            arr[j] = temp
        gap //= 2 # Уменьшение шага вдвое на каждой итерации


if __name__ == "__main__":
    try:
        a = input('a = ').split()
        if not a:
            numbers_to_sort = []
        else:
            numbers_to_sort = [int(x) for x in a]
    except ValueError:
        print("Ошибка: Введите числа, разделенные пробелами.")
        numbers_to_sort = []

    shell_sort(numbers_to_sort)
    print('Ожидается:', numbers_to_sort)
```
