Задача 18. Полиномиальный хэш с модулем.

Добавить поддержку модуля q для
предотвращения переполнения.


Вычислите хеш по формуле: s[0]*31^(n-1) + s[1]*31^(n-2) + ... результат брать по модулю 109+7109+7

```python
def poly_hash_mod(s: str, q: int = 10**9 + 7) -> int:
    """
    Полиномиальный хеш с модулем q для предотвращения переполнения.
    s: входная строка
    q: модуль (по умолчанию 10^9 + 7)
    :return: хеш-значение
    Предотвращение переполнения — на каждом шаге цикла результат берётся по модулю q,
    поэтому числа никогда не превышают q * 31 ≈ 3 * 10^10, что безопасно для Python
    Гибкость — можно передавать любой модуль: 7225, 10^9+7, 10^9+9 и т.д.
    10**9+7 — это самое большое простое число, помещающееся в 32-битный знаковый тип данных,
    которое позволяет выполнять быстрые операции умножения и взятия остатка, не прибегая к длинной арифметике, что важно для скорости.
    """
    n = len(s)
    result = 0
    for i in range(n):
        # На каждом шаге берём остаток от деления на q,
        # чтобы числа не росли экспоненциально и не происходило переполнение
        result = (result + ord(s[i]) * pow(31, n - 1 - i, q)) % q
    return result
    ''' ord(s[i]) — получает числовой код символа
        Формула: s[0]*31^(n-1) + s[1]*31^(n-2) + ... + s[n-1]*31^0
        используется pow(31, степень, MOD) для эффективного возведения в степень по модулю
        '''

if __name__ == '__main__':
    # Тестовые строки
    test_strings = ["abc", "hello", "test", "a"]

    MOD = 109 + 7109 + 7
    MOD_STANDARD = 10**9 + 7   # Стандартный модуль 10^9 + 7

    print(f"Хеш с модулем q = {MOD} (109+7109+7):")
    for s in test_strings:
        print(f"poly_hash_mod('{s}', q) = {poly_hash_mod(s, MOD)}")
    print()

    print(f"Хеш с модулем q = {MOD_STANDARD} (10^9+7):")
    for s in test_strings:
        print(f"poly_hash_mod('{s}', q) = {poly_hash_mod(s, MOD_STANDARD)}")
    print()

    # Пример с длинной строкой
    print("Пример с длинной строкой:")
    long_str = "polynomial_hash_example_1234567890"
    print(f"poly_hash_mod('{long_str}', q={MOD}) = {poly_hash_mod(long_str, MOD)}")
    print(f"poly_hash_mod('{long_str}', q={MOD_STANDARD}) = {poly_hash_mod(long_str, MOD_STANDARD)}")
```
